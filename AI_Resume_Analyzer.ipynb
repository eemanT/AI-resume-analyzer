```python
def analyze_resume(resume_text, job_description):

    # -----------------------------
    # 1. Find skills in resume
    # -----------------------------
    resume_lower = resume_text.lower()

    found_skills = []

    for skill in skills_list:
        if skill.lower() in resume_lower:
            found_skills.append(skill)

    # -----------------------------
    # 2. Find skills in job
    # -----------------------------
    job_lower = job_description.lower()

    required_skills = []

    for skill in skills_list:
        if skill.lower() in job_lower:
            required_skills.append(skill)

    # -----------------------------
    # 3. Compare skills
    # -----------------------------
    matched_skills = []
    missing_skills = []

    for skill in required_skills:
        if skill in found_skills:
            matched_skills.append(skill)
        else:
            missing_skills.append(skill)

    # -----------------------------
    # 4. Skill score
    # -----------------------------
    if len(required_skills) > 0:
        skill_score = (
            len(matched_skills) / len(required_skills)
        ) * 100
    else:
        skill_score = 0

    # -----------------------------
    # 5. AI semantic similarity
    # -----------------------------
    resume_embedding = model.encode(resume_text)
    job_embedding = model.encode(job_description)

    similarity = cosine_similarity(
        [resume_embedding],
        [job_embedding]
    )[0][0]

    match_percentage = similarity * 100

    # -----------------------------
    # 6. Final score
    # -----------------------------
    final_score = (
        match_percentage * 0.60
    ) + (
        skill_score * 0.40
    )

    # -----------------------------
    # 7. Match level
    # -----------------------------
    if final_score >= 80:
        match_level = "Excellent Match"
    elif final_score >= 65:
        match_level = "Good Match"
    elif final_score >= 50:
        match_level = "Moderate Match"
    else:
        match_level = "Weak Match"

    # -----------------------------
    # 8. AI feedback
    # -----------------------------
    feedback_prompt = f"""
You are an AI career assistant.

Analyze this candidate's resume against the job description.

RESUME:
{resume_text}

JOB DESCRIPTION:
{job_description}

MATCH SCORE:
{final_score:.1f}%

MATCHED SKILLS:
{matched_skills}

MISSING SKILLS:
{missing_skills}

Give:

1. Resume Strengths
2. Missing Skills
3. Improvement Suggestions
4. Overall Recommendation

Keep the answer practical and concise.
"""

    feedback_result = generator(
        feedback_prompt,
        max_new_tokens=400
    )

    ai_feedback = feedback_result[0]["generated_text"]

    # -----------------------------
    # 9. Interview questions
    # -----------------------------
    interview_prompt = f"""
You are an AI interview coach.

Generate 5 interview questions based on this
resume and job description.

RESUME:
{resume_text}

JOB DESCRIPTION:
{job_description}

MATCHED SKILLS:
{matched_skills}

MISSING SKILLS:
{missing_skills}

Include:
1. Two technical questions
2. One project question
3. One missing-skill question
4. One behavioral question

Number them from 1 to 5.
"""

    interview_result = generator(
        interview_prompt,
        max_new_tokens=300
    )

    interview_questions = interview_result[0]["generated_text"]

    # -----------------------------
    # 10. Return everything
    # -----------------------------
    return {
        "match_score": round(final_score, 1),
        "match_level": match_level,
        "matched_skills": matched_skills,
        "missing_skills": missing_skills,
        "ai_feedback": ai_feedback,
        "interview_questions": interview_questions
    }

print("Resume analyzer function created successfully!")
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Remove the old FLAN model from GPU memory
try:
    del model
    del tokenizer
    torch.cuda.empty_cache()
except:
    pass

model_name = "Qwen/Qwen2.5-1.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("New AI model loaded successfully!")
print("Model:", model_name)
print("Device:", next(model.parameters()).device)
```

```python
def qwen_generate(instruction, max_new_tokens=200):
    messages = [
        {
            "role": "system",
            "content": "You are a professional AI resume analyzer. Follow the user's instructions exactly."
        },
        {
            "role": "user",
            "content": instruction
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=4096
    ).to(model.device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.15,
            pad_token_id=tokenizer.eos_token_id
        )

    # Only decode the newly generated part
    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    return tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()


def generate_feedback(resume_text, job_description):
    prompt = f"""
Compare this resume with the job description.

RESUME:
{resume_text[:3000]}

JOB DESCRIPTION:
{job_description[:2500]}

Write exactly 4 short sentences:
1. Mention the strongest skills that match the job.
2. Mention the most important missing skill.
3. Mention one relevant project or experience.
4. Give one specific improvement recommendation.

Do not copy the resume word-for-word.
"""

    return qwen_generate(prompt, max_new_tokens=150)


def generate_interview_questions(resume_text, job_description):
    prompt = f"""
Create exactly 5 different technical interview questions for this candidate.

RESUME:
{resume_text[:3000]}

JOB DESCRIPTION:
{job_description[:2500]}

Requirements:
- Exactly 5 questions.
- Every question must be different.
- Do not ask the candidate to name a project.
- Questions should relate to the candidate's skills, projects, and the job.
- Number them 1 to 5.
- Return ONLY the questions.
"""

    return qwen_generate(prompt, max_new_tokens=200)


print("Qwen resume-analysis functions are ready!")
```

```python
def generate_interview_questions(resume_text, job_description):
    prompt = f"""
Generate exactly 5 technical interview questions using ONLY facts explicitly
contained in the resume and job description.

RESUME:
{resume_text[:3000]}

JOB DESCRIPTION:
{job_description[:2500]}

STRICT RULES:
- Exactly 5 questions.
- Number them 1 to 5.
- Every question must be different.
- Do not invent experience, technologies, responsibilities, or achievements.
- Do not assume the candidate did something merely because they list a skill.
- Ask about what the candidate actually built, used, or learned.
- At least 3 questions must refer directly to actual projects.
- Questions should be appropriate for a Junior Software Engineer.
- Keep each question concise, preferably under 25 words.
- Do not add explanations or answers.
- Output ONLY the 5 numbered questions.

Good examples:
- "How did you combine CNN, LSTM, and BiLSTM in your cryptocurrency prediction project?"
- "How did you use ANN and fuzzy logic together in your diabetes detection system?"
- "How did you use FastAPI in your cryptocurrency prediction application?"

Bad examples:
- Do not ask about PyTorch if it is not mentioned.
- Do not ask about Flask if it is not mentioned.
- Do not assume real-time optimization if it is not mentioned.
- Do not assume the candidate performed an activity just because a technology
  appears in their skills list.
"""

    return qwen_generate(prompt, max_new_tokens=250)


questions = generate_interview_questions(
    resume_text,
    job_description
)

print("=" * 70)
print("🎤 INTERVIEW QUESTIONS")
print("=" * 70)
print(questions)
```

```python
# ==========================================
# STEP 21 — SKILL MATCHING
# ==========================================

def find_skill_matches(resume_text, job_description):

    common_skills = [
        "Python",
        "JavaScript",
        "React",
        "FastAPI",
        "Flutter",
        "HTML",
        "CSS",
        "SQL",
        "Git",
        "TensorFlow",
        "Keras",
        "Machine Learning",
        "Deep Learning",
        "Artificial Intelligence",
        "Fuzzy Logic",
        "ANN",
        "LSTM",
        "CNN",
        "BiLSTM"
    ]

    resume_lower = resume_text.lower()
    job_lower = job_description.lower()

    resume_skills = []
    job_skills = []
    matched_skills = []

    for skill in common_skills:
        skill_lower = skill.lower()

        if skill_lower in resume_lower:
            resume_skills.append(skill)

        if skill_lower in job_lower:
            job_skills.append(skill)

        if skill_lower in resume_lower and skill_lower in job_lower:
            matched_skills.append(skill)

    return {
        "resume_skills": resume_skills,
        "job_skills": job_skills,
        "matched_skills": matched_skills
    }


skill_analysis = find_skill_matches(
    resume_text,
    job_description
)

print("=" * 70)
print("🔎 SKILL MATCH ANALYSIS")
print("=" * 70)

print("\nSkills found in resume:")
print(", ".join(skill_analysis["resume_skills"]))

print("\nSkills mentioned in job:")
print(", ".join(skill_analysis["job_skills"]))

print("\n✅ Matching skills:")
print(", ".join(skill_analysis["matched_skills"]))
```

```python
# ==========================================
# STEP 22 — RESUME / JOB MATCH SCORE
# ==========================================

def calculate_match_score(skill_analysis):

    job_skills = skill_analysis["job_skills"]
    matched_skills = skill_analysis["matched_skills"]

    if len(job_skills) == 0:
        return 0

    score = (len(matched_skills) / len(job_skills)) * 100

    return round(score, 2)


match_score = calculate_match_score(skill_analysis)

print("=" * 70)
print("📊 RESUME-JOB MATCH SCORE")
print("=" * 70)

print(f"\nMatch Score: {match_score}%")

if match_score >= 80:
    print("🟢 Strong Match")

elif match_score >= 60:
    print("🟡 Good Match")

elif match_score >= 40:
    print("🟠 Moderate Match")

else:
    print("🔴 Low Match")
```

```python
# ==========================================
# STEP 23 — FINAL COMBINED ANALYSIS
# ==========================================

def run_full_analysis(resume_text, job_description):

    # 1. Find skills
    skill_analysis = find_skill_matches(
        resume_text,
        job_description
    )

    # 2. Calculate match score
    match_score = calculate_match_score(
        skill_analysis
    )

    # 3. Generate AI feedback
    feedback = generate_feedback(
        resume_text,
        job_description
    )

    # 4. Generate interview questions
    interview_questions = generate_interview_questions(
        resume_text,
        job_description
    )

    # 5. Store everything together
    final_result = {
        "match_score": match_score,
        "resume_skills": skill_analysis["resume_skills"],
        "job_skills": skill_analysis["job_skills"],
        "matched_skills": skill_analysis["matched_skills"],
        "ai_feedback": feedback,
        "interview_questions": interview_questions
    }

    return final_result


# Run complete analysis
final_result = run_full_analysis(
    resume_text,
    job_description
)

print("=" * 70)
print("📄 FINAL RESUME ANALYSIS")
print("=" * 70)

print(f"\n📊 Match Score: {final_result['match_score']}%")

print("\n✅ MATCHED SKILLS:")
print(", ".join(final_result["matched_skills"]))

print("\n🤖 AI FEEDBACK:")
print(final_result["ai_feedback"])

print("\n🎤 INTERVIEW QUESTIONS:")
print(final_result["interview_questions"])

print("\n" + "=" * 70)
print("✅ ANALYSIS COMPLETE")
print("=" * 70)
```
